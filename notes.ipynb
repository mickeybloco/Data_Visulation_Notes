### Matplotlib Functions  
-  `plt.plot()` - default creaates line chart (trends over time)  
- `plt.show()` - starts eventloop, that opens interactive window to view active figs (similar to print but for plots)    
- `plt.scatter()` - scatter plot, visualizing relations  
- `plt.bar()` - comparing categorical data  
- `plt.barh()` - hz bar  
- `plt.hist()` - barplot for continuous data  
- `plt.savefig()` - save a plot to a file  
- `plt.contour()` - draws contour lines
- `plt.annotate()` - create annotations on the plot  




#### Plot Vocabulary:  
- figure (fig) - container, window to put your plots on  
    - figsize  
    - dpi - dots per inch  
    - face color  
    - subplotspaars - distance between subplots  
- axes (ax) - area within the fig where data is plotted  
- axis - x,y axis of a plot or chart  
- line - data itself, cant be fully customized  
- legend - small container with labels/descriptors  
- labels - names/labels for the x and y axis  
- title - title for chart/plot  
- xlabel
- markers - symbols that represent data points   
- grid  - background (attribute that can be edited)   
- ticks(tic) - marks along the x/y/axis

```python
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 170, 190, 160, 300, 310, 320, 330])
```

```python
plt.plot(
    x, y,
    linestyle='dotted', 
    color='green', 
    marker='D', 
    markerfacecolor='purple', 
    markersize=12
    )

plt.title('Sports Watch Data')
plt.xlabel('Average Pulse')
plt.ylabel('Calories Burned')

plt.grid(color='red', linestyle=':', alpha=0.7)
plt.gca().set_facecolor('lavender')   # inside the graph
plt.gcf().set_facecolor('lightyellow')   # outside the graph
for i in range(len(x)):
    plt.text(
        x[i],
        y[i] + 5,
        str(y[i]),
        ha='center',
        fontweight='bold'
    )
plt.show()
```

```python
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(1, 2, 1) # create 1 row, with 2 plots, this plot is index 1
plt.plot(x, y)
plt.title('Plot 1')
x1 = np.array([0, 1, 2, 3])
y1 = np.array([10, 20, 25, 30])

plt.subplot(1, 2, 2)# create 1 row, with 2 plots, this plot is index 2
plt.title('Plot 2')
plt.plot(x1, y1)
plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y4 = np.exp(x)

fig, axes = plt.subplots(2, 2)

axes[0, 0].plot(x, y1)
axes[0, 0].set_title('Sin Function')

axes[0, 1].plot(x, y2)
axes[0, 1].set_title('Cosine Function')

axes[1, 0].plot(x, y3)
axes[1, 0].set_title('Tangent Function')

axes[1, 1].plot(x, y4)
axes[1, 1].set_title('Exponential Function')

plt.tight_layout()
plt.show()

```

```python
years=[1950,1955,1960,1965,1970,1980,1985,1990,1995,2000,2005,2010,2015]
pops=[2.5,2.7,3.0,3.3,3.6,4.0,4.4,4.8,5.3,5.7,6.1,6.5,7.3]
death=[1.2,1.1,1.2,2.1,2.0,2.3,1.8,1.9,2.6,1.6,2.4,2.4,4.0]
job_rate = [0.5, 0.6, 0.8, 0.9, 1.0, 1.2, 1.5, 1.7, 2.0, 2.2, 2.5, 2.7, 2.9]
```

```python
plt.plot(years, pops, color='red', linestyle='--', label='Population')
plt.plot(years, death, color='green', linestyle=':', label='Death')
plt.plot(years, job_rate, color='purple', linestyle='-.', label='Jobs')
plt.title('Population Growth')
plt.xlabel('Population Growth by Year')
plt.ylabel('Population(Billions)')
plt.legend()
plt.grid(alpha=0.3, linestyle='--')
plt.show()
```
