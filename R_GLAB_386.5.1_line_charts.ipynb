```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
x = np.arange(1, 12, 2)
y = np.arange(2, 13, 2)

# create a line plot with a dashdot linestyle
plt.plot(x, y, linestyle='-.', marker='*', markersize='20', color='green')
plt.show()
```

```python
fig, ax = plt.subplots()

ax.plot(x, y, linestyle='-.', marker='*', color='purple')
plt.show()
```

```python
plt.plot(x, y, linestyle=':')
plt.show()
```

```python
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 170, 190, 160, 300, 310, 320, 330])
```

```python
plt.plot(
    x, y,
    linestyle='dotted', 
    color='green', 
    marker='D', 
    markerfacecolor='purple', 
    markersize=12
    )

plt.title('Sports Watch Data')
plt.xlabel('Average Pulse')
plt.ylabel('Calories Burned')
plt.show()
```

```python
plt.plot(
    x, y,
    linestyle='dotted', 
    color='#FF5733', 
    marker='D', 
    markerfacecolor='purple', 
    markersize=12
    )

plt.title('Sports Watch Data')
plt.xlabel('Average Pulse', color='pink', size=18, style='italic')
plt.ylabel('Calories Burned', color=(0.5, 0.5, 0.7))

plt.show()
```
